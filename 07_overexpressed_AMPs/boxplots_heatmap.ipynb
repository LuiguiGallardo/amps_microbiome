```python
data = pd.read_csv("./amps_overexpressed_ob_groups.tsv", delimiter="\t", index_col=0)

# data["group"] = ["obese", "obese", "obese", "oms", "oms", "oms"]

samples_cols = ["O 1", "O 2", "O 3", "OMS 1", "OMS 2", "OMS 3"]
data[samples_cols] = data[samples_cols].apply(pd.to_numeric, errors='coerce')

print(data)

my_colors = ["orange", "red"]
sns.set_palette(my_colors)

# Column colors
group = data.pop("overexpressed_group")
print(group)
column_colors = ["orange", "red", "blue", "green"]
unique_values = group.unique()[:len(column_colors)]
lut = dict(zip(unique_values, column_colors))
col_colors = group.map(lut)

# Row colors
origin = data.pop("origin")
row_colors = ["purple", "green", "yellow", "red"]
unique_values = origin.unique()[:len(row_colors)]
lut = dict(zip(unique_values, row_colors))
row_colors = origin.map(lut)

heatmap = sns.clustermap(
    np.log10(data[samples_cols]),
    # col_colors=col_colors,
    row_colors=row_colors,
    xticklabels=True,
    yticklabels=True,
    cmap="BuPu",
    cbar_kws={"label": "log10(Abundance)"}
)
# plt.xlabel("Samples")
# plt.ylabel("AMPs")
plt.tick_params(labelsize=6)
plt.tight_layout()
plt.savefig("./heatmap_log10_ob_groups.svg")
plt.savefig("./heatmap_log10_ob_groups.png")
plt.clf()
```
