```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

def plot_genome_coverage_from_file(file_path, title="Genome Coverage Line Plot"):
    """
    Plots a line plot of genome coverage per base from a TSV file.

    Parameters:
        file_path (str): Path to the TSV file containing the data.
        title (str): The title of the plot.
    """
    # Read the TSV file into a DataFrame
    df = pd.read_csv(file_path, sep="\t")
    
    # Plot the line plot using Seaborn
    plt.figure(figsize=(12, 6))
    sns.lineplot(x="Position", y="Coverage", data=df, marker="o", color="blue")
    plt.title(title)
    plt.xlabel("Position")
    plt.ylabel("Coverage")
    plt.grid(axis='y', linestyle='--', alpha=0.7)
    plt.show()

# Example usage
file_path = "../data/coverage_data.tsv"
plot_genome_coverage_from_file(file_path)
```
